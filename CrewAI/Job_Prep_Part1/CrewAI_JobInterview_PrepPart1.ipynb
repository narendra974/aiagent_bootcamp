```python
# Define the interview prep task

interview_prep_task = Task(
    description = f"""
    Interview the user for the job {job_position} with the following description: {job_description}
    """,
    agent = interviewer_agent,
    expected_output = f"""
    ask only one question to the user that is relevant for the job position {job_position}
    """, 
    human_input = True
)
```

# AI CREW

```python
crew = Crew(
    name = "Interview Prep Crew",
    tasks = [interview_prep_task],
    agents = [interviewer_agent],
    verbose = True,
    processes = Process.sequential
)
results = await crew.kickoff_async()
print(results)
```
